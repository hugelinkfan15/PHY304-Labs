```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def euler_newton(force, times, xI, vI, m):
    xA = np.zeros(times.size)
    vA = np.zeros(times.size)
    xA[0] = xI
    vA[0] = vI
    for i in range(0,times.size-1):
        dt = times[i+1]-times[i]
        a = force(times[i], xA[i], vA[i]) / m
        vA[i+1] = vA[i] + a*dt
        xA[i+1] = xA[i] + vA[i+1]*dt
    return xA, vA
```

mg = -bv

**-mg/b = v**

mg = -cv|v|

**-mg/c|v| =  v**

**Linear Air Resistance**

*m = 10g, b = 2*

```python
def linR(t,x,v):
    return -9.8 - 2*v
```

```python
t = np.linspace(0,30,50)
xL, vL = euler_newton(linR, t, 0, 0, 10)
```

```python
plt.plot(t,xL)
plt.xlabel("Time")
plt.ylabel("Position")
plt.title("Linear Drag Time vs Position")
plt.tight_layout()
plt.grid()
plt.show()
```

```python
plt.plot(t,vL)
plt.xlabel("Time")
plt.ylabel("Velocity")
plt.title("Linear Drag Time vs Velocity")
plt.tight_layout()
plt.grid()
plt.show()
```

**Quadratic Air Resistance**
*m=10, c= 2*

```python
def quadR(t,x,v):
    return -9.8- 2*v*abs(v)
```

```python
t = np.linspace(0,30,50)
xV, vV = euler_newton(quadR, t, 0, 0, 10)
```

```python
plt.plot(t,xV)
plt.xlabel("Time")
plt.ylabel("Velocity")
plt.title("Quadratic Drag Time vs Velocity")
plt.tight_layout()
plt.grid()
plt.show()
```

```python
plt.plot(t,vV)
plt.xlabel("Time")
plt.ylabel("Velocity")
plt.title("Quadratic Drag Time vs Velocity")
plt.tight_layout()
plt.grid()
plt.show()
```

When comparing the linear and quadratic drag velocity graphs, it seems that the quadratic drag approaches terminal velocity more rapidly than the linear drag.
