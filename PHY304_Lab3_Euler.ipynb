```python
import numpy as np
import matplotlib.pyplot as plt
import os
os.getcwd()
```

```python
#1.1
def euler_newton(force, times, xI, vI, m):
# create arrays for x and v
# store the initial conditions in the first elements
    xA = np.zeros(times.size)
    vA = np.zeros(times.size)
    xA[0] = xI
    vA[0] = vI
    for i in range(0,times.size-1):
        dt = times[i+1]-times[i]
        a = force(times[i], xA[i], vA[i]) / m
        vA[i+1] = vA[i] + a*dt
        xA[i+1] = xA[i] + vA[i+1]*dt
    return xA, vA
```

```python
def gravity(t,x,v):
    return -9.8 
```

```python
t = np.linspace(0,5,50)
print(t,"\n")
x,v = euler_newton(gravity,t, 10, 0, 1)
```

```python
plt.plot(t,x)
plt.xlabel("time")
plt.ylabel("distance")
plt.tight_layout()
plt.grid()
plt.show()
```

```python
def ana_X (xI,vI,times):
    xA = np.zeros(times.size)
    vA = np.zeros(times.size)
    xA[0] = xI
    vA[0] = vI
    for i in range(1,times.size-1):
        xA[i+1]= xI+(vI*times[i])-(0.5*9.8*times[i]*times[i])
        vA[i+1]= vI-9.8*times[i]
    return xA, vA
```
