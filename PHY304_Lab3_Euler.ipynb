```python
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
#1.1
def euler_newton(force, times, xI, vI, m):
# create arrays for x and v
# store the initial conditions in the first elements
    xA = np.zeros(times.size)
    vA = np.zeros(times.size)
    xA[0] = xI
    vA[0] = vI
    for i in range(0,times.size-1):
        dt = times[i+1]-times[i]
        a = force(times[i], xA[i], vA[i]) / m
        vA[i+1] = vA[i] + a*dt
        xA[i+1] = xA[i] + vA[i+1]*dt
    return xA, vA
```

```python
def gravity(t,x,v):
    return -9.8 
```

```python
t = np.linspace(0,3,50)
x,v = euler_newton(gravity,t, 10, 0, 1)
```

```python
plt.plot(t,x)
plt.xlabel("time")
plt.ylabel("distance")
plt.tight_layout()
plt.grid()
plt.show()
```

```python
def ana_X (xI,vI,times):
    xA = np.zeros(times.size)
    vA = np.zeros(times.size)
    xA[0] = xI
    vA[0] = vI
    for i in range(1,times.size):
        xA[i]= xI+(vI*times[i])-(0.5*9.8*times[i]*times[i])
        vA[i]= vI-9.8*times[i]
    return xA, vA
```

**3) A Simple Harmonic Oscillator**

```python
def spring(k,t,x,v):
    return -k*x
```

```python
tS = np.linspace(0,60,600)
xS,vS = euler_newton(lambda t,x,v: spring(1,t,x,v),tS,1, 0, 1)
```

```python
plt.plot(tS,xS)
plt.xlabel("Time (s)")
plt.ylabel("Position (m)")
plt.title("Spring Oscillator")
plt.tight_layout()
plt.grid()
plt.show()
```

```python
plt.plot(tS,vS)
plt.xlabel("Time (s)")
plt.ylabel("Velocity (m/s)")
plt.title("Spring Oscillator")
plt.tight_layout()
plt.grid()
plt.show()
```

```python
plt.plot(tS,np.cos(tS))
plt.xlabel("Time (s)")
plt.ylabel("Position (m)")
plt.title("Spring Oscillator")
plt.tight_layout()
plt.grid()
plt.show()
```

```python
plt.plot(tS,-np.sin(tS))
plt.xlabel("Time (s)")
plt.ylabel("Velocity (m/s)")
plt.title("Spring Oscillator")
plt.tight_layout()
plt.grid()
plt.show()
```

**1)** The simulated oscillator have the expected period of 1 rad/s.  
**2)** The simulated oscillator seems to follow the analytical solution quite closey, with not much difference observable to the naked eye.  
**3)** As *$\Delta$t* grows smaller, the simulated oscillator begins to deviated from the analytical solution more and more.  
**4)** The evidence I have is that my simulated results appear to line up with those expected from the analytical results.
